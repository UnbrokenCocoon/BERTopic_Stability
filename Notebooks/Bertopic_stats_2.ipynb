```python
import pickle
with open(os.path.join(data_dir, r'uc_sentences.pkl'), 'rb') as f:
  all_sentences= pickle.load(f)
```

```python
import pandas as pd
df = pd.read_csv(os.path.join(data_dir, r'uc_bertopic_eval_results.csv'))
```

```python
print(df)
```

```python
sort_metrics = df[['Error_Size', 'Keyword_Freq_Score']]
sort_metrics['Appearance Percentage'] =(sort_metrics['Error_Size']/len(all_sentences))*100
print(sort_metrics)
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(10, 6))
plt.hist(sort_metrics['Appearance Percentage'], bins=30, color='skyblue', edgecolor='black')
plt.title("Distribution of Appearance Percentage Small Corpus")
plt.xlabel("Appearance Percentage")
plt.ylabel("Frequency")
plt.grid(axis='y')
plt.show()
```

```python
from scipy.stats import spearmanr
import matplotlib.pyplot as plt
from scipy import stats
sort_metrics = df.sort_values(by='Error_Size', ascending=True)
sort_metrics = sort_metrics[['Error_Size', 'Gini_Score']]
sort_metrics = sort_metrics.drop_duplicates()
def myfunc(x):
  return slope * x + intercept


x = sort_metrics['Error_Size']
y = sort_metrics['Gini_Score']

slope, intercept, r, p, std_err = stats.linregress(x, y)

mymodel = list(map(myfunc, x))
plt.xlabel("Error_Size")
plt.ylabel("Gini_Score")

plt.scatter(x, y)
plt.plot(x, mymodel)
rho, p_value = spearmanr(x, y)
print(f"Spearman correlation = {rho:.2f}, p = {p_value:.4f}")
plt.show()
```
